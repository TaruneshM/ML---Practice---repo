```python
from sklearn.datasets import make_regression as mr 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.linear_model import LinearRegression 
```

```python
x,y=mr(n_samples=4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
plt.scatter(x,y)
```

```python
reg=LinearRegression()
reg.fit(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color="red")
```

```python
# lets apply gradient decent where m=78.35 and lets take b=0  
y_pred=((78.35*x)+0).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='OLS')
plt.plot(x,y_pred,color='#00a65a',label='b = 0')
plt.legend()
plt.show()
```

```python
m=78.35
lr=0.1
b=0
epoch=100

for i in range (epoch):
    ls=-2*np.sum(y-m*x.ravel()-b)
    b=b-(lr*ls)
    y_pred=m*x +b
    plt.plot(x,y_pred)
plt.scatter(x,y)
```

```python
b = 0
m = 78.35
lr = 0.01

epochs = 10

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*x.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * x + b

  plt.plot(x,y_pred)

plt.scatter(x,y)
     
```
